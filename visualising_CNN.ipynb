```python
from keras.applications.vgg16 import VGG16
#load the model
model = VGG16()
#summarize the model
model.summary() # it is a 16 layered network
```

```python
from keras.utils import plot_model

plot_model(model, to_file='vgg.png')
```

```python
for i in range(len(model.layers)):
  #checking for convulution layer
  if 'conv' not in model.layers[i].name:
    continue

  #get filter weights
  filters, biases = model.layers[i].get_weights()
  print("layer number",i,model.layers[i],filters.shape)
```

```python
# retrive weights from second hidden layer
filter ,bias =model.layers[1].get_weights()
```

```python
#normalize filter values to 0-1 so we can visualize them
f_min,f_max = filters.min(),filters.max()
filters = (filters - f_min) / (f_max - f_min)
```

```python
import matplotlib.pyplot as pyplot
n_filters =6
ix=1
fig = pyplot.figure(figsize=(15,10))
for i in range(n_filters):
  #get the filter
  f = filters[:,:,:,i]

  for j in range(3):
    # subplot for 6 filters and 3 channels
    pyplot.subplot(n_filters,3,ix)
    pyplot.imshow(f[:,:,j],cmap='gray')
    ix+=1
#show the figure
pyplot.show()
```

```python
from keras.models import Model
model = Model(inputs=model.inputs, outputs=model.layers[i].output)
```

```python
from keras.preprocessing.image import load_img, img_to_array
from numpy import expand_dims
from keras.applications.vgg16 import preprocess_input

img = load_img('/content/VK.jpeg',target_size=(224,224))

#convert img to array
img = img_to_array(img)
#expand dimensions so that it represents a single 'sample'
img = expand_dims(img,axis=0)
img = preprocess_input(img)
```

```python
#calculating features_map
features =model.predict(img)

fig = pyplot.figure(figsize=(20,15))
# Determine how many feature maps to display, limited by the grid size (8x8 = 64)
num_to_display = min(features.shape[3], 64)

for i in range(1, num_to_display + 1):

  pyplot.subplot(8,8,i)
  pyplot.imshow(features[0,:,:,i-1],cmap='gray')

pyplot.show()
```

```python
model2 =VGG16()
```

```python
layer_index =[2,5,9,13,17]
outputs =[model2.layers[i].output for i in layer_index]

model3= Model(inputs=model2.inputs,outputs=outputs)
```

```python
feature_map =model3.predict(img)

for layer_idx in range(len(feature_map)):
  fig = pyplot.figure(figsize=(20,15))
  # Use the actual layer number for the title
  fig.suptitle("Feature Maps for Layer {}".format(layer_index[layer_idx]), fontsize=20)

  # Get the current layer's feature map output
  current_layer_features = feature_map[layer_idx]
  num_filters_in_layer = current_layer_features.shape[3]

  # Determine how many feature maps to display, limited by the grid size (8x8 = 64)
  num_to_display = min(num_filters_in_layer, 64)

  for filter_idx in range(num_to_display):
    pyplot.subplot(8,8,filter_idx + 1) # subplot indices are 1-based
    pyplot.imshow(current_layer_features[0,:,:,filter_idx],cmap='gray')

  pyplot.show()
```
