```python
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator #used tensor flow backend
```

```python
img = image.load_img('/content/Left Cat.jpg',target_size=(200,200))
```

```python
img
```

```python
type(img)
```

```python
#making object of the class called datagen #no vertical flip ,fill mode is soething cool like ,reflect,constant etc
datagen = ImageDataGenerator(rotation_range=40,width_shift_range=0.2,height_shift_range=0.2,shear_range=0.2,zoom_range=0.2,horizontal_flip=True,fill_mode='nearest')
```

```python
img = image.img_to_array(img)
```

```python
type(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1,200,200,3)
```

```python
import os

i=0

if not os.path.exists('aug'):
  os.makedirs('aug')

for output in datagen.flow(input_batch,batch_size=1,save_to_dir='aug'):

  i= i+1

  if i == 10:
    break
```
